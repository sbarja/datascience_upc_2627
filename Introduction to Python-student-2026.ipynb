## 2. Variables and data structures

Variables in Python are created once they are defined, that is, when a value is assigned to the variable for the first time (using the <code> = </code> operator). For this reason, variables (and their type) do not need to be declared in Python.

## 3. Conditional structures (if)

Conditional control structures allow you to execute one part of the code or another depending on the evaluation of one or several boolean conditions of YES or NO (**<code>TRUE</code>** or **<code>FALSE</code>**).

In Python, conditional control structures are defined by the words **<code>if</code>**, **<code>elif</code>** and **<code>else</code>**.

- **<code>if CONDITION:</code>** if the conditional expression is met, the following code block is executed.
- **<code>elif CONDITION:</code>** otherwise, if this conditional expression is met, this other code block is executed.
- **<code>else:</code>** otherwise, this code block is executed without evaluating any condition.


The condition is usually evaluated by the following relational operators <code> <, <=, ==, >=, >, != </code>.

The part of code inside an **<code>if</code>** cannot be empty, but if for some reason an if statement must be created without content, the **<code>pass</code>** statement can be used to prevent an error from occurring.

````python
# Example

celsius = -5
fahrenheit = 9.0 /5.0 * celsius + 32

print("Temperature in Fahrenheit is", fahrenheit)

if fahrenheit > 90:
    print("It is hot")
elif fahrenheit < 30:
    print("It is cold")
else: print("It is neither hot nor cold")
````

## 4. Lists and loops

### 4.1 Iterative control structures (for & while)

Iterative control structures (also called loops) allow the same code to be executed repeatedly as long as a condition is met.

There are two types of iterative control structures:

* For
* While

The Python ``for`` statement iterates over the elements of any sequence (a list or a string for example), in the order they appear in the sequence.

The ``while`` loop allows multiple iterations to be performed based on the result of a logical expression that can result in a ``True`` or ``False`` value.

````python
# Example of for

for i in range(0,10,1):
    print(i)
````

````python
# Example of while

i=0
while i < 10 :
    print(i)
    i= i+1
````

In addition, loops with ``for`` or ``while`` can be combined with conditional ``if`` structures to create more complex structures.

With the ``break`` statement we can stop the loop before it has gone through all elements.

With the ``continue`` statement we can stop the current iteration of the loop (in both ``for`` and ``while``) and continue with the next one.

````python
# Example

numbers = [-5, 3, 2, -1, 9, 6]
sum_positive_1 = 0

for n in numbers:
    if n >= 0:
        sum_positive_1 += n

sum_positive_2 = 0
for n in numbers:
    if n < 0:
        continue
    sum_positive_2 += n

print(sum_positive_1, sum_positive_2)
````

### 4.2 Lists (or dictionary) comprehensions

List comprehensions are a way of fitting a ``for`` loop, an ``if`` statement, and an assignment all on a single line.

A list comprehension consists of the following parts:

+ An input sequence
+ A variable representing members of the input sequence
+ An optional expression
+ An output expression that produces elements of the output list from members of the input sequence that satisfy the predicate

````python
# Example
num = [1, 4, -5, 10, -7, 2, 3, -1]
squared = []
for i in num:
     if i > 0:
        squared.append(i**2)
print(type(squared), squared)

# This for can be rewritten as follows:
# In an iteration, it first checks if the if is TRUE and then performs the calculation
squared = [x**2 for x in num if x > 0] 
print(type(squared), squared)
````

## 6. Functions
A function is a block of code with an associated name, which receives zero or more arguments as input, follows a sequence of statements where the desired operations are executed and returns a value and/or performs a task. This code block can be called multiple times as needed.

In Python there are a number of functions built into the language by default, but you can also create user-defined functions to use in your own programs. 

The complete list of default built-in functions can be found at:
https://docs.python.org/3/library/functions.html

````python
# Examples of built-in functions in Python
print(len([1, 2, 3]))       # length
print(max(3, 7, 2))         # maximum
print(round(3.14159, 2))    # round to 2 decimals
print(abs(-5))              # absolute value
````

To create a function in Python, you must use the word ``def`` to define the function. The syntax for a function definition in Python is:

<code>*def* function_name(input_parameter_list):
    function_statements
    *return* [expression_to_return]</code>

Where:
* function_name: is the name of the function.
* input_parameter_list: is the list of parameters that a function can receive, separated by comma.
* function_statements: is the block of statements in Python source code that perform a given operation.
* expression_to_return: is the expression or variable that returns the return statement.

````python
# Example: function to set the first value of a list to 0

def change_first_element(values):
    values[0] = 0   # no return: the function modifies the list it receives (lists are mutable)

numbers = [1, 2, 3, 4]
change_first_element(numbers)
print(numbers)
````

#### Example: capacity factor

The capacity factor of a power plant is the ratio between the energy it actually produced over a period and the energy it would have produced running at nominal power the whole time:

$$CF = \frac{E}{P_{nom} \cdot h}$$

The text between triple quotes is a *docstring*: it documents what the function does (place the cursor on the function name and press `Shift+Tab` to see it).

````python
def capacity_factor(energy_mwh, power_mw, hours):
    """Return the capacity factor (0-1) of a plant."""
    return energy_mwh / (power_mw * hours)

# A 10 MW wind farm that produced 26,280 MWh in one year (8760 h)
capacity_factor(26280, 10, 8760)
````

### 6.1 Arguments and parameters of functions

When defining a function, the values that are received are called parameters, while during the call, the values that are sent are called arguments.

By default, a function must be called with the correct number of arguments. If the function expects 2 arguments, you must call the function with 2 arguments.

#### Arguments per position

When arguments are sent to a function, by default they are received in order in the defined parameters, known as arguments by position. In the example above, the first argument (26280) goes to `energy_mwh`, the second (10) to `power_mw` and the third (8760) to `hours`.

#### Arguments by name

However, it is possible to skip the order of the parameters if you indicate during the call what value each parameter has from its name.

````python
capacity_factor(hours=8760, power_mw=10, energy_mwh=26280)
````

#### Default parameters

A parameter can have a default value, which is used when the argument is not given in the call. Parameters with a default value go after those without one.

````python
def capacity_factor(energy_mwh, power_mw, hours=8760):
    """Return the capacity factor (0-1). By default, over one year (8760 h)."""
    return energy_mwh / (power_mw * hours)

print(capacity_factor(26280, 10))            # uses hours=8760
print(capacity_factor(600, 10, hours=720))   # one month
````

#### Indeterminate parameters ``*args`` and ``**kwargs`` (optional)

Sometimes it is not known in advance how many items need to be sent to a function.

If you do not know how many arguments per item will be passed to a function, you add a * before the parameter name in the function definition. By convention, the parameter name ``*args`` is usually used. In this way, the function will receive a tuple of arguments and can access the items accordingly.

Similarly, if it is not known how many arguments by name will be passed to a function, two ** are added before the parameter name in the function definition. By convention, the parameter name ``**kwargs`` is usually used. In this way, the function will receive a dictionary of arguments and can access the elements accordingly.

````python
def super_function(*args,**kwargs):
    total = 0
    for arg in args:
        total += arg
    print ("sum => ", total)
    for kwarg in kwargs:
        print (kwarg, "=>", kwargs[kwarg])

super_function(50, -1, 1.56, 10, 20, 300, name="Peter", age=38)
````

### 6.2 Lambda functions

Sometimes we need a small function only once, typically to pass it as an argument to another function. Instead of defining it with `def`, we can write an anonymous or *lambda* function in a single line:

``lambda inputs: output``

````python
square = lambda x: x**2      # equivalent to:  def square(x): return x**2
square(4)
````

Their most common use is as an argument of another function. For example, ``sorted`` accepts an optional ``key`` function that tells it *how* to sort (https://docs.python.org/3/library/functions.html#sorted):

````python
technologies = ['wind', 'Solar', 'hydraulic', 'Nuclear', 'biomass', 'Coal']
print(sorted(technologies))                            # uppercase letters go first!
print(sorted(technologies, key=lambda x: x.lower()))   # case-insensitive
````

## 7. Objects and methods (optional)

Almost everything in Python is an **object**: a bundle of data (*attributes*) and functions that act on that data (*methods*). You have used them already: `text.lower()` or `my_list.append(4)` call a method of a string or a list object.

A **class** is the template used to create objects, and each object created from it is an *instance* of the class. Classes are defined with the word ``class``. The special method ``__init__`` runs when an object is created and stores its attributes in ``self`` (the object itself).

<img src="Figures/clases-vEng.png" alt="Drawing" style="width: 300px;"/>

````python
class PVPlant:
    """A simple photovoltaic plant."""

    def __init__(self, name, peak_power_kw, efficiency=0.85):
        self.name = name                    # attributes
        self.peak_power_kw = peak_power_kw
        self.efficiency = efficiency

    def power(self, irradiance):            # method
        """AC power (kW) for a given irradiance (W/m2). Reference irradiance: 1000 W/m2."""
        return self.peak_power_kw * irradiance / 1000 * self.efficiency

plant = PVPlant('Rooftop ETSEIB', peak_power_kw=100)
print(plant.name, plant.peak_power_kw)
print(plant.power(800), 'kW')
````

The array object in NumPy is called ``ndarray``, it provides many support functions that make working with ``ndarray`` very easy. You can create an ndarray with NumPy with the ``.array()`` function.

### 8.2 Why NumPy: vectorization

Operations on NumPy arrays are applied to all the elements at once (**vectorization**), without writing loops. Besides being shorter, it is much faster, because the loop runs in compiled code instead of Python. Let's convert one year of 15-minute power readings from kW to MW and compare (`%timeit` measures how long a line takes):

````python
power_kw = np.random.default_rng(0).uniform(0, 5000, size=35_040)   # one year of 15-min readings
power_list = list(power_kw)

# Python list: we need a loop
%timeit [p / 1000 for p in power_list]

# NumPy: the operation is applied to the whole array
%timeit power_kw / 1000
````

**Broadcasting:** operations between an array and a single number, or between arrays of compatible shapes, are extended automatically to all elements. For example, the power (kW) of 3 plants during 4 time steps of 15 minutes:

````python
power = np.array([[100, 120,  90],     # rows: time steps, columns: plants (kW)
                  [110, 130,  95],
                  [ 80, 125, 100],
                  [ 60, 110,  85]])

energy_kwh = power * 0.25              # the number is applied to every element
price = np.array([0.10, 0.12, 0.11])  # €/kWh, one price per plant: applied to every row

print(energy_kwh)
print(energy_kwh * price)
````

````python
import plotly.express as px

fig = px.line(meter_113, x='day', y=['energy_max', 'energy_mean', 'energy_min'],
              title='Summary MAC000113 consumption', labels={'value': 'Energy (kWh)', 'day': 'Date'})
fig.show()
````
